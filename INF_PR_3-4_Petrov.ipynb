#### Задание 1. Бит чётности

##### Ручной расчёт

101011 - 4 единицы -> бит чётности = 0

тогда кодовое слово: 1010110

##### Программа

```python
def even_bit_encode(msg):
    ones = msg.count("1")
    print(f"Кол-во единиц: {ones}")
    pb = ones % 2
    print(f"Бит чётности: {pb}")
    msg_encoded = f"{msg}{pb}"
    print(f"Кодированное сообщение: {msg_encoded}")
    return msg_encoded


def even_bit_valid(emsg):
    epb = int(emsg[-1])
    msg = emsg[:-1]
    ones = msg.count("1")
    print(f"Кол-во единиц: {ones}")
    pb = ones % 2
    print(f"Ожидаемый бит чётности: {pb}; а в кодированном сообщении: {epb}")
    if epb == pb:
        print("Чётность соблюдена, сообщение верное")
        return True
    print("Чётность не соблюдена, сообщение неверное")
    return False
```

```python
message = "101011"
encoded_message = even_bit_encode(message)
even_bit_valid(encoded_message)
```

Ломаем 1 бит в закодированном сообщение

```python
arr_enc_msg = list(encoded_message)
arr_enc_msg[1] = "1"
break_encoded_message = "".join(arr_enc_msg)
print(f"Сломанное сообщение: {break_encoded_message}")
even_bit_valid(break_encoded_message)
```

Ломаем ещё один бит (в итоге сломано 2)

```python
arr_enc_msg[0] = "0"
break_encoded_message = "".join(arr_enc_msg)
print(f"Сломанное сообщение: {break_encoded_message}")
even_bit_valid(break_encoded_message)
```

Хотя чётность и соблюдена - сообщение все равно не верное (не равно исходному) т.к. проверка четности не будет работать если изменено чётное количество бит

#### Задание 3. Троирование

##### Ручной расчёт

Сообщение: 1001 - 4 бита

Троированное: 111 000 000 111 - 12 бит

Вырос в 3 раза


Сломанное: 101 000 000 111

Восстановим: 111 000 000 111


##### Программа

```python
def triple_encode(msg):
    res = ""
    for s in msg:
        res += 3 * s
    return res
```

```python
message = "1011010100"
message_encoded = triple_encode(message)
message_encoded
```

ломаем код

```python
em_arr = list(message_encoded)
em_arr[3] = "1"
message_encoded = "".join(em_arr)
message_encoded
```

```python
def triple_decode(emsg):
    res = ""
    arr = [emsg[i:i+3] for i in range(0, len(emsg), 3)]
    for els in arr:
        if els.count("1") >= 2:
            res += "1"
        else:
            res += "0"
    return res

msg_decoded = triple_decode(message_encoded)
msg_decoded
```

#### Задание 6. Программная реализация

```python
def hamming74_encode(data_bits):
    d1, d2, d3, d4 = (int(bit) for bit in data_bits)

    p1 = d1 ^ d2 ^ d4
    p2 = d1 ^ d3 ^ d4
    p4 = d2 ^ d3 ^ d4

    return f"{p1}{p2}{d1}{p4}{d2}{d3}{d4}"
```

```python
def hamming74_syndrome(code_bits):
    b = [int(bit) for bit in code_bits]

    s1 = b[0] ^ b[2] ^ b[4] ^ b[6]
    s2 = b[1] ^ b[2] ^ b[5] ^ b[6]
    s4 = b[3] ^ b[4] ^ b[5] ^ b[6]

    return 4 * s4 + 2 * s2 + s1
```

```python
def hamming74_correct(code_bits):
    b = [int(bit) for bit in code_bits]
    position = hamming74_syndrome(code_bits)

    if position != 0:
        b[position - 1] ^= 1

    return "".join(str(bit) for bit in b)
```

```python
def hamming74_decode(code_bits):
    fixed = hamming74_correct(code_bits)
    return fixed[2] + fixed[4] + fixed[5] + fixed[6]
```

```python
data = "1101"
code = hamming74_encode(data)

print("Исходные данные:", data)
print("Код Хэмминга:", code)

for i in range(7):
    broken = list(code)
    print(f"\nДелаем ошибку в позиции: {i + 1}")
    broken[i] = "0" if broken[i] == "1" else "1"
    broken = "".join(broken)

    syndrome = hamming74_syndrome(broken)
    fixed = hamming74_correct(broken)
    decoded = hamming74_decode(broken)

    print("Принято с ошибкой:", broken)
    print("Синдром:", syndrome)
    print("Исправленное слово:", fixed)
    print("Восстановленные данные:", decoded)
```
