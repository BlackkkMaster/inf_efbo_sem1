#### Задание 1. Бит чётности

##### Ручной расчёт

101011 - 4 единицы -> бит чётности = 0

тогда кодовое слово: 1010110

##### Программа

```python
def even_bit_encode(msg):
    ones = msg.count("1")
    print(f"Кол-во единиц: {ones}")
    pb = ones % 2
    print(f"Бит чётности: {pb}")
    msg_encoded = f"{msg}{pb}"
    print(f"Кодированное сообщение: {msg_encoded}")
    return msg_encoded


def even_bit_valid(emsg):
    epb = int(emsg[-1])
    msg = emsg[:-1]
    ones = msg.count("1")
    print(f"Кол-во единиц: {ones}")
    pb = ones % 2
    print(f"Ожидаемый бит чётности: {pb}; а в кодированном сообщении: {epb}")
    if epb == pb:
        print("Чётность соблюдена, сообщение верное")
        return True
    print("Чётность не соблюдена, сообщение неверное")
    return False
```

```python
message = "101011"
encoded_message = even_bit_encode(message)
even_bit_valid(encoded_message)
```

Ломаем 1 бит в закодированном сообщение

```python
arr_enc_msg = list(encoded_message)
arr_enc_msg[1] = "1"
break_encoded_message = "".join(arr_enc_msg)
print(f"Сломанное сообщение: {break_encoded_message}")
even_bit_valid(break_encoded_message)
```

Ломаем ещё один бит (в итоге сломано 2)

```python
arr_enc_msg[0] = "0"
break_encoded_message = "".join(arr_enc_msg)
print(f"Сломанное сообщение: {break_encoded_message}")
even_bit_valid(break_encoded_message)
```

Хотя чётность и соблюдена - сообщение все равно не верное (не равно исходному) т.к. проверка четности не будет работать если изменено чётное количество бит

#### Задание 2. Расстояние Хэмминга

##### Ручной расчёт

|сообщение A|1|1|0|1|0|1|0|1|
|---|---|---|---|----|----|---|---|---|
|сообщение B|1|0|1|0|0|1|1|1|
|---|---|---|---|----|----|---|---|---|
|отличие||+|+|+|||+||

размер сообщения: 8 бит

расстояние: 4

##### Программаная реализация

```python
msgA = "11010101"
msgB = "10100111"
d = 0
for i in range(len(msgA)):
    if msgA[i] != msgB[i]:
        d += 1

d
```

##### Объяснение

d=4 показыват, что в сравнении с msgA, msgB имеет расстояние Хэмминга = 4, т.е. различается в 3 битах. Если msgB - полученное с ошибками сообщение, а msgA один из эталонов (исходных сообщений, коды которых мы знаем), то сравнивая дальше msgB с различными эталонами msgX, и вычисляя для каждого значение d, а затем взяв эталон msgX с самым маленьким d, мы получим наиболее вероятное исходное слово (т.е. msgB = msgX)

#### Задание 3. Троирование

##### Ручной расчёт

Сообщение: 1001 - 4 бита

Троированное: 111 000 000 111 - 12 бит

Вырос в 3 раза


Сломанное: 101 000 000 111

Восстановим: 111 000 000 111


##### Программа

```python
def triple_encode(msg):
    res = ""
    for s in msg:
        res += 3 * s
    return res
```

```python
message = "1011010100"
message_encoded = triple_encode(message)
message_encoded
```

ломаем код

```python
em_arr = list(message_encoded)
em_arr[3] = "1"
message_encoded = "".join(em_arr)
message_encoded
```

```python
def triple_decode(emsg):
    res = ""
    arr = [emsg[i:i+3] for i in range(0, len(emsg), 3)]
    for els in arr:
        if els.count("1") >= 2:
            res += "1"
        else:
            res += "0"
    return res

msg_decoded = triple_decode(message_encoded)
msg_decoded
```
